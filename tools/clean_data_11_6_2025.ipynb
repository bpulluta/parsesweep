```python
def fix_state_abbr_in_address(address):
    return (
        address
        .replace(", VA ", ", Virginia ")
        .replace(", OH ", ", Ohio ")
        .replace(", IL ", ", Illinois ")
        .replace(", IN ", ", Indiana ")
        .replace(", NJ ", ", New Jersey ")
        .replace(", MD ", ", Maryland ")
        .replace(", DE ", ", Delaware ")
        .replace(", NC ", ", North Carolina ")
        .replace(", PA ", ", Pennsylvania ")
        .replace(", KY ", ", Kentucky ")
        .replace(", MI ", ", Michigan ")
        .replace(", WV ", ", West Virginia ")
        .replace(", TN ", ", Tennessee ")
    )

data['facility_address'] = data['facility_address'].apply(fix_state_abbr_in_address)
```

# Fix county names

```python
def fix_county(county_name):
    if pd.isnull(county_name):
        return county_name

    if not county_name:
        return None

    county_name = (
        county_name
        .replace('(Fairfax County)', 'Fairfax County')
        .replace(' and City of Manassas', '')
        .replace(', New Jersey', '')
        .replace(', MD', '')
        .lstrip().rstrip()
        .removesuffix(" County")
    )

    if "Prince George" not in county_name:
        county_name = county_name.title()

    return county_name

data["facility_county"] = data["facility_county"].apply(fix_county)
```

# Add missing generator data

```python
data["manual_capacity_lookup"] = False
gen_info = pd.read_csv("generators_missing_capacity_V2.csv")
for __, row in gen_info.iterrows():
    mask = (
        (data["permit_number"] == row["permit_number"])
        & (data["make"] == row["make"])
        & (data["model"] == row["model"])
    )
    if not mask.any():
        continue

    if not pd.isnull(row["rated_capacity_hp"]):
        data.loc[mask, "rated_capacity_hp"] = float(row["rated_capacity_hp"])
    if not pd.isnull(row["rated_capacity_bhp"]):
        data.loc[mask, "rated_capacity_bhp"] = float(row["rated_capacity_bhp"])
    if not pd.isnull(row["rated_capacity_kw"]):
        data.loc[mask, "rated_capacity_kw"] = float(row["rated_capacity_kw"])
    if not pd.isnull(row["rated_capacity_mmbtu_per_hr"]):
        data.loc[mask, "rated_capacity_mmbtu_per_hr"] = float(row["rated_capacity_mmbtu_per_hr"])
    data.loc[mask, "manual_capacity_lookup"] = True
```
